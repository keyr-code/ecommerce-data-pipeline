```python
import boto3

s3 = boto3.client(
    's3',
    endpoint_url='http://localhost:9000',
    aws_access_key_id='rustfsadmin',
    aws_secret_access_key='rustfsadmin'
)
```

```python
# Create bucket
s3.create_bucket(Bucket='bucket-test')
```

```python
# List buckets
print(s3.list_buckets())
```

```python
#Upload file
s3.put_object(Bucket='bucket-test', Key='test.txt', Body=b'Hello RustFS')
```

```python
# Download file
obj = s3.get_object(Bucket='bucket-test', Key='test.txt')
print(obj['Body'].read())
```
